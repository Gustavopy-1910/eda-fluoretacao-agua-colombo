```python
import pandas as pd

# Carregando a planilha
df = pd.read_excel('fluoreto_colombo_PR.xlsx', sheet_name='Formato_longo')

# Exibindo as 5 primeiras linhas da tabela
df.head()

# Exibindo as informações estruturais do DataFrame (linhas, colunas e tipos de dados)
df.info()

# Calculando a média global de flúor
media_geral = df['fluoreto_ppm'].mean()

# Calculando o desvio-padrão global de flúor
desvio_geral = df['fluoreto_ppm'].std()

print("Média geral:", media_geral)
print("Desvio-padrão geral:", desvio_geral)

# Agrupando os dados por Unidade de Saúde e calculando a média e o desvio-padrão de flúor
resumo_unidades = df.groupby('unidade')['fluoreto_ppm'].agg(['mean', 'std'])

# Exibindo a tabela com o resumo de cada unidade
resumo_unidades

import matplotlib.pyplot as plt
import seaborn as sns

# Criando o histograma da concentração de flúor
sns.histplot(data=df, x='fluoreto_ppm', kde=True)

# Personalizando o gráfico com títulos e rótulos
plt.title('Distribuição da Concentração de Flúor em Colombo/PR')
plt.xlabel('Flúor (ppm)')
plt.ylabel('Frequência (Quantidade de Medições)')

# Exibindo o gráfico
plt.show()

import matplotlib.pyplot as plt
import seaborn as sns

# Ajustando o tamanho da figura para facilitar a leitura das 15 unidades
plt.figure(figsize=(14, 6))

# Criando o Boxplot por unidade de saúde
sns.boxplot(data=df, x='unidade', y='fluoreto_ppm')

# Adicionando linhas horizontais com os limites da legislação (0.60 e 0.80 ppm)
plt.axhline(0.60, color='red', linestyle='--', label='Mínimo ideal (0.60 ppm)')
plt.axhline(0.80, color='red', linestyle='--', label='Máximo ideal (0.80 ppm)')

# Rotação dos rótulos do eixo X para não sobrepor os nomes
plt.xticks(rotation=45)
plt.title('Comparativo da Concentração de Flúor entre as 15 Unidades de Saúde')
plt.xlabel('Unidade de Saúde')
plt.ylabel('Flúor (ppm)')
plt.legend()
plt.show()

import pandas as pd

# 1. Agrupar os dados por unidade e calcular média e desvio-padrão
resumo = df.groupby('unidade')['fluoreto_ppm'].agg(
    media='mean',
    desvio_padrao='std',
    total_medicoes='count'
).reset_index()

# 2. Identificar unidades críticas (fora do intervalo 0.60 - 0.80 ppm)
unidades_criticas = resumo[
    (resumo['media'] < 0.60) | (resumo['media'] > 0.80) | (resumo['desvio_padrao'] > 0.20)
].sort_values(by='media', ascending=False)

# Exibir as unidades críticas no Colab
print("--- UNIDADES COM INDICADORES CRÍTICOS ---")
display(unidades_criticas)

# 3. Exportar o resultado para um arquivo Excel
unidades_criticas.to_excel('relatorio_unidades_criticas.xlsx', index=False)
print("\nArquivo 'relatorio_unidades_criticas.xlsx' gerado com sucesso!")
```
